```python
import os
import wikipedia
from langchain_openai import ChatOpenAI

# Configurar el idioma de Wikipedia
wikipedia.set_lang("es")

# Configuración del LLM
try:
    llm = ChatOpenAI(
        model="gpt-4o",
        openai_api_base=os.environ.get("GITHUB_BASE_URL"),
        openai_api_key=os.environ.get("GITHUB_TOKEN"),
        temperature=0
    )
    print("✅ LLM de LangChain configurado.")
except Exception as e:
    print(f"❌ Error configurando el LLM: {e}")
    llm = None

from langchain_classic.agents import tool, create_openai_tools_agent, AgentExecutor
from langsmith import Client

@tool
def get_wikipedia_summary(query: str) -> str:
    """Busca en Wikipedia un tema y devuelve un resumen de 2 frases. Útil para obtener información sobre personas, lugares o conceptos."""
    try:
        return wikipedia.summary(query, sentences=2)
    except Exception as e:
        return f"Ocurrió un error: {e}"

tools = [get_wikipedia_summary]

client = Client(None)
# Usamos el mismo prompt de la comunidad que ya está preparado para manejar historial
prompt = client.pull_prompt("hwchase17/openai-tools-agent")

agent = create_openai_tools_agent(llm, tools, prompt)
agent_executor = AgentExecutor(agent=agent, tools=tools, verbose=True)

print("✅ Agente y herramientas listos.")
```

## 2. Responsabilidad Organizacional

Las organizaciones que desarrollan agentes de IA tienen responsabilidades críticas:

```python
class OrganizationalResponsibility:
    """Framework para gestionar responsabilidades organizacionales en IA"""
    
    def __init__(self):
        self.compliance_frameworks = [
            "EU AI Act",
            "GDPR",
            "CCPA",
            "SOC 2"
        ]
        self.security_measures = []
        self.ethical_guidelines = []
    
    def add_security_measure(self, measure):
        """Agregar medida de seguridad"""
        self.security_measures.append(measure)
        print(f"✓ Medida de seguridad agregada: {measure}")
    
    def add_ethical_guideline(self, guideline):
        """Agregar guía ética"""
        self.ethical_guidelines.append(guideline)
        print(f"✓ Guía ética agregada: {guideline}")
    
    def check_compliance(self, framework):
        """Verificar cumplimiento con un framework específico"""
        if framework in self.compliance_frameworks:
            return f"Framework {framework} reconocido"
        return f"Framework {framework} no implementado"
    
    def generate_report(self):
        """Generar reporte de estado"""
        return {
            "compliance_frameworks": self.compliance_frameworks,
            "security_measures_count": len(self.security_measures),
            "ethical_guidelines_count": len(self.ethical_guidelines)
        }

# Ejemplo de uso
org_responsibility = OrganizationalResponsibility()
org_responsibility.add_security_measure("Input validation")
org_responsibility.add_security_measure("Output filtering")
org_responsibility.add_ethical_guideline("No generar contenido dañino")
org_responsibility.add_ethical_guideline("Proteger privacidad de usuarios")

print("\n📊 Reporte de Responsabilidad Organizacional:")
print(org_responsibility.generate_report())
```
